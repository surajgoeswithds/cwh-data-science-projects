```python
with open('initialdata.txt' , encoding = 'utf-8' ) as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n")
chunks
```

```python
chunks = [ c for c in chunks if len(c) > 3 ]
chunks
```

```python
chunks[4]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    lines = chunk.split('\n')
    username = lines[0]
    posts = lines[1].replace(" posts" , "").replace(" post" , "")
    followers = lines[2].replace(" followers" , "")
    following = lines[3].replace(" following" , "")
    bio = "\n".join(lines[4:])

    return {"username": username , "posts": posts , "followers": followers , "following": following , "bio": bio}
    
```

```python
parse_chunk(chunks[4])
```

```python
print(parse_chunk(chunks[4])['bio'])
```

```python
profiles = [parse_chunk(c) for c in chunks] #builds a NEW LIST by applying a FUNCTION to EACH ITEM of chunks
```

```python
#profiles is now in a FORMAT that makes ANSWERING THE PROJECT'S ACTUAL QUESTIONS possible
```

```python
profiles[0]
```

```python
max = 0
for profile in profiles:
    posts_num = int(profile['posts'].replace(',', ''))
    if (posts_num > max):
        max = posts_num
        profile_with_max_posts = profile
f"details of the person with maximum posts:{profile_with_max_posts}"
```

```python
def convert_followers(value):
    if value.endswith("K"):
        return float(value.replace("K", "")) * 1000
    else:
        return int((value).replace(',' , ''))

max_follower = 0
for profile in profiles:
    followers_num = convert_followers(profile['followers'])
    if (followers_num > max_follower):
        max_follower = followers_num
        profile_with_max_followers = profile
f'details of person with maximum followers:{profile_with_max_followers}'
```
