```python
import json
def load(filename):
    with open(filename , 'r') as f:
        return json.load(f)
```

```python
def find_pages_you_might_like(user_id, data):
    user_likedpages = {}
    for user in data['users']:
        user_likedpages[user['id']] = set(user['liked_pages'])
    likedpagesby_user = user_likedpages[user_id]
    #Why does this matter(this is for for loop): you're trying to compare the target user against EVERY OTHER user. user_likedpages contains all 30 users' data. likedpagesby_user contains only ONE user's (the target's) pages.
    page_suggestions = {}
    for other_user , pages in user_likedpages.items():
        if other_user != user_id:
            shared_pages = likedpagesby_user.intersection(pages)
            if shared_pages: #checked FIRST. Recall — an empty set is FALSY, basically if score=0 then doesnt suggest pages
                for suggest_page in pages:  #suggest_page (a single page id)
                    if suggest_page not in likedpagesby_user:
                        page_suggestions[suggest_page] = page_suggestions.get(suggest_page, 0) + len(shared_pages)
    sorted_pages = sorted(page_suggestions.items() , key=lambda x: x[1] , reverse = True)
    return [page_id for page_id, _ in sorted_pages]
```

```python
data = load('massive_data2.json')
user_id = 7
page_recommendations = find_pages_you_might_like(user_id , data)
print(f"Pages You Might Like for User {user_id}: {page_recommendations}")
```
